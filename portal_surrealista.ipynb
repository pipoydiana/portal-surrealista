```python
import gradio as gr
from huggingface_hub import InferenceClient

import gradio as gr
from huggingface_hub import InferenceClient

# Intentamos cargar el token de forma segura desde los Secretos de Colab
try:
    from google.colab import userdata
    HF_TOKEN = userdata.get('HF_TOKEN')
except Exception:
    HF_TOKEN = "TU_TOKEN_DE_HUGGING_FACE_AQUI" # Respaldo por si se ejecuta fuera de Colab

# El cliente usa esa variable de forma privada
cliente = InferenceClient("stabilityai/stable-diffusion-xl-base-1.0", token=HF_TOKEN)

lista_atmosferas = [
    "un desierto de arenas líquidas y relojes derretidos",
    "un bosque de cristal fragmentado brillando en la oscuridad",
    "un océano de nubes flotantes bajo un cielo verde esmeralda",
    "una ciudad infinita hecha completamente de espejos",
    "una habitación sin gravedad llena de esferas de agua flotantes"
]

def transmutar_texto(objeto, atmosfera):
    if not objeto: raise gr.Error("Por favor, ingresa un objeto cotidiano.")
    prompt = f"A surrealist painting of a {objeto}, melting into {atmosfera}, Salvador Dali and Rene Magritte style, dreamlike, impossible geometry, museum exhibition lighting, 8k masterpiece"
    try:
        return cliente.text_to_image(prompt)
    except Exception as e:
        raise gr.Error(f"Error del servidor: {str(e)}")

def transmutar_imagen(foto, atmosfera):
    if foto is None: raise gr.Error("Por favor, sube o toma una foto.")
    prompt = f"A high-end surrealist art piece inspired by a real object, melting into {atmosfera}, Salvador Dali and Rene Magritte style, dreamlike, museum exhibition lighting, 8k"
    try:
        return cliente.text_to_image(prompt)
    except Exception as e:
        raise gr.Error(f"Error del servidor: {str(e)}")


# --- DISEÑO DE GALERÍA DE ARTE / MUSEO (CSS AVANZADO) ---

tema_museo = gr.themes.Soft(
    primary_hue="purple",
    neutral_hue="slate",
    font=[gr.themes.GoogleFont("Playfair Display"), "sans-serif"]
)

estilo_museo = """
/* Fondo general estilo sala de exhibición */
body {
    background-color: #0f0e13 !important;
}

/* Cabecera del Museo */
#galeria-header {
    text-align: center;
    padding: 20px;
    background: linear-gradient(135deg, #1a1823 0%, #0f0e13 100%);
    border-bottom: 1px solid #2d2a3d;
    margin-bottom: 25px;
    border-radius: 12px;
}
#titulo-museo {
    font-family: 'Playfair Display', serif;
    color: #e2d9fc;
    font-size: 2.8em;
    letter-spacing: 2px;
    margin-bottom: 5px;
}
#subtitulo-museo {
    color: #9a93af;
    font-style: italic;
    font-size: 1.1em;
}

/* Efecto de marco de cuadro para las imágenes generadas */
.output-image img {
    border: 8px solid #2a2638 !important;
    box-shadow: 0 15px 35px rgba(0,0,0,0.7) !important;
    border-radius: 4px !important;
}
"""

with gr.Blocks(theme=tema_museo, css=estilo_museo) as demo:

    # Encabezado con estética de galería
    with gr.Column(elem_id="galeria-header"):
        gr.HTML("<h1 id='titulo-museo'>GALERÍA ONÍRICA</h1>")
        gr.HTML("<p id='subtitulo-museo'>Exhibición Interactiva de Transmutación Surrealista — Portal de Arte Contemporáneo e Inteligencia Artificial</p>")

    with gr.Tabs():
        # SALA 1: Transmutación por Texto
        with gr.TabItem("🏛️ Sala I: Objeto Conceptual (Texto)"):
            with gr.Row():
                with gr.Column(scale=1):
                    gr.Markdown("### 📝 Configuración de la Obra")
                    entrada_objeto = gr.Textbox(
                        label="1. Objeto Cotidiano",
                        placeholder="Ej. Una llave antigua, un zapato, una manzana..."
                    )
                    entrada_atmosfera_txt = gr.Dropdown(
                        choices=lista_atmosferas,
                        label="2. Dimensión o Atmósfera Onírica",
                        value=lista_atmosferas[0]
                    )
                    btn_texto = gr.Button("🖼️ Exponer Obra en la Sala", variant="primary")

                with gr.Column(scale=1):
                    gr.Markdown("### 🖼️ Lienzo del Museo")
                    resultado_texto = gr.Image(type="pil", label="Resultado de la Exposición", elem_classes="output-image")

            btn_texto.click(fn=transmutar_texto, inputs=[entrada_objeto, entrada_atmosfera_txt], outputs=resultado_texto)

        # SALA 2: Transmutación por Fotografía
        with gr.TabItem("📷 Sala II: Registro Fotográfico (Imagen)"):
            with gr.Row():
                with gr.Column(scale=1):
                    gr.Markdown("### 📸 Registro del Objeto Real")
                    entrada_foto = gr.Image(type="pil", label="Sube o fotografía tu objeto físico")
                    entrada_atmosfera_img = gr.Dropdown(
                        choices=lista_atmosferas,
                        label="2. Dimensión o Atmósfera Onírica",
                        value=lista_atmosferas[0]
                    )
                    btn_foto = gr.Button("✨ Transmutar Fotografía al Óleo", variant="primary")

                with gr.Column(scale=1):
                    gr.Markdown("### 🖼️ Lienzo del Museo")
                    resultado_foto = gr.Image(type="pil", label="Obra Surrealista Transmutada", elem_classes="output-image")

            btn_foto.click(fn=transmutar_imagen, inputs=[entrada_foto, entrada_atmosfera_img], outputs=resultado_foto)

if __name__ == "__main__":
    demo.launch(share=True, debug=True)
```
